### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
a_25 = np.array([1, 2])
b_25 = np.array([2, 1])

dot_product = np.dot(a_25, b_25)
norm_a = np.linalg.norm(a_25)
norm_b = np.linalg.norm(b_25)

cos_angle = dot_product / (norm_a * norm_b)
angle_rad = np.arccos(cos_angle)
angle_deg = np.degrees(angle_rad)

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_deg)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A_33 = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

means = A_33.mean(axis=0)
stds = A_33.std(axis=0)
normalized_A = (A_33 - means) / stds

print("Нормалізована матриця (по стовпцях):\n", normalized_A)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A_40 = np.array([
    [1,  2,  1],
    [2,  1, -1],
    [3, -1,  2]
])
b_40 = np.array([9, 8, 3])

x_40 = np.linalg.solve(A_40, b_40)
print("Вектор розв'язку x:", x_40)

check = A_40 @ x_40
print("Перевірка A @ x:", check)
print("Вектор b:", b_40)
print("Результати збігаються:", np.allclose(check, b_40))
````
